## With memory

```python
# import the builder from the studio/agent_with_memory.py
from studio.agent_with_memory import builder
from langgraph.checkpoint.memory import InMemorySaver

react_agent_with_memory = builder.compile(checkpointer=InMemorySaver())
```

```python
# Invoke the first and second questions again.

ra1 = react_agent_with_memory.invoke(input_messages, config=config)

ra1["messages"][-1].pretty_print()
```

```python

ra2 = react_agent_with_memory.invoke(input_messages_2, config=config)

ra2["messages"][-1].pretty_print()
```

```python
input_messages_3 = {
    "messages": [
        {"role":"user", "content":"當月的最後一天？"}
    ]
}

ra3 = react_agent_with_memory.invoke(input_messages_3, config=config)

ra3["messages"][-1].pretty_print()
```

```python
# print all messages in ra3
for m in ra3["messages"]:
    m.pretty_print()
    
```
