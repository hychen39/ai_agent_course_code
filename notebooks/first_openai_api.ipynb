```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

chat_model = init_chat_model("gpt-5-nano")
```

```python
# invoke the chat model with a prompt
prompt_str = "Hi, please introduce yourself."
# AIMessage object
response = chat_model.invoke(prompt_str)
```

```python
type(response)  # <class 'langchain_core.messages.ai.AIMessage'>
```

```python
print("Response from the chat model:", response)
```

```python
print(response.content)
```
