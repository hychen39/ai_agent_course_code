# Memory Store 的基本操作方法

```python
# Create a InMemoryStore instance
from langgraph.store.memory import InMemoryStore

store = InMemoryStore()
```

```python
# Save the user_123's profile
store.put(
    namespace=("users", "user_123", "preferences"),
    key="communication",
    value={
        "preferred_language": "台灣繁體中文",
        "preferred_answer_style": "精簡、直接"
    }
)
```

```python
# Save the user_456's profile
store.put(
    namespace=("users", "user_456", "preferences"),
    key="communication",
    value={
        "preferred_language": "English",
        "preferred_answer_style": "detailed, step-by-step"
    }
)
```

## Read items from a namespace

```python
# Search for all items in the namespace ("users", "user_123", "preferences")
user_123_preferences = store.search(("users", "user_123", "preferences"))

for item in user_123_preferences:
    print(f"Key: {item.key}, Value: {item.value}")
```

```python
# List items in multiple namespaces
# 提供部份的 namespace_prefix, 取得所有符合的資料, 使用 search
namespace_prefix= ("users",)

users_all_pref = store.search(namespace_prefix, limit=10)
for item in users_all_pref:
    print(f"Namespace: {item.namespace}, Key: {item.key}, Value: {item.value}")
```

```python
# 取得 `user_123` 的偏好資料, 使用 get

user_123_pref_comm = store.get(
    namespace=("users", "user_123", "preferences"),
    key="communication"
)

print(user_123_pref_comm)
```
